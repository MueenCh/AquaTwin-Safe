```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
import numpy as np
import pandas as pd

# Show labels first
print("Dataset 1 labels:")
print(train_normal["ATT_FLAG"].value_counts(dropna=False))

print("\nDataset 2 labels:")
print(train_attack["ATT_FLAG"].value_counts(dropna=False))

# Common sensor columns
feature_cols = [
    col for col in test_data.columns
    if col != "DATETIME"
]

# Dataset 1 is normal training data
X_train = train_normal[feature_cols].copy()

# Use only labelled rows from Dataset 2
validation = train_attack[
    train_attack["ATT_FLAG"].isin([1, -1])
].copy()

X_valid = validation[feature_cols].copy()

# BATADAL: 1 = normal, -1 = attack
y_valid = validation["ATT_FLAG"].map({
    1: 0,
    -1: 1
}).astype(int)

# Handle invalid/missing values
X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_valid = X_valid.replace([np.inf, -np.inf], np.nan)

training_medians = X_train.median()
X_train = X_train.fillna(training_medians)
X_valid = X_valid.fillna(training_medians)

# Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_valid_scaled = scaler.transform(X_valid)

# Isolation Forest
model = IsolationForest(
    n_estimators=300,
    contamination=0.05,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train_scaled)

# Isolation Forest: -1 anomaly, 1 normal
raw_predictions = model.predict(X_valid_scaled)
y_pred = (raw_predictions == -1).astype(int)

print("\nEvaluation results")
print("Accuracy :", round(accuracy_score(y_valid, y_pred), 4))
print("Precision:", round(precision_score(y_valid, y_pred, zero_division=0), 4))
print("Recall   :", round(recall_score(y_valid, y_pred, zero_division=0), 4))
print("F1-score :", round(f1_score(y_valid, y_pred, zero_division=0), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_valid, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_valid,
    y_pred,
    target_names=["Normal", "Attack"],
    zero_division=0
))
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split

tf.random.set_seed(42)
np.random.seed(42)

# Chronological split of normal Dataset 1
split_index = int(len(X_train_scaled) * 0.80)

X_ae_train = X_train_scaled[:split_index]
X_ae_validation = X_train_scaled[split_index:]

print("Autoencoder training samples:", X_ae_train.shape)
print("Autoencoder validation samples:", X_ae_validation.shape)

number_of_features = X_ae_train.shape[1]

autoencoder = Sequential([
    Input(shape=(number_of_features,)),
    Dense(32, activation="relu"),
    Dropout(0.10),
    Dense(16, activation="relu"),
    Dense(8, activation="relu", name="latent_space"),
    Dense(16, activation="relu"),
    Dense(32, activation="relu"),
    Dense(number_of_features, activation="linear")
])

autoencoder.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mse"
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=7,
    restore_best_weights=True
)

history = autoencoder.fit(
    X_ae_train,
    X_ae_train,
    validation_data=(X_ae_validation, X_ae_validation),
    epochs=50,
    batch_size=64,
    shuffle=True,
    callbacks=[early_stopping],
    verbose=1
)

autoencoder.summary()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.title("BATADAL Autoencoder Training")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    r"D:\BATADAL\results\autoencoder_training_loss.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Reconstruction errors for normal validation data
validation_reconstruction = autoencoder.predict(
    X_ae_validation,
    verbose=0
)

validation_errors = np.mean(
    np.square(X_ae_validation - validation_reconstruction),
    axis=1
)

# Reconstruction errors for Dataset 2
attack_reconstruction = autoencoder.predict(
    X_valid_scaled,
    verbose=0
)

attack_errors = np.mean(
    np.square(X_valid_scaled - attack_reconstruction),
    axis=1
)

# Evaluate predefined thresholds
percentiles = [95, 97.5, 99]
autoencoder_results = []

for percentile in percentiles:
    threshold = np.percentile(validation_errors, percentile)
    predictions = (attack_errors > threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_valid,
        predictions,
        labels=[0, 1]
    ).ravel()

    autoencoder_results.append({
        "Model": "Autoencoder",
        "Threshold_Percentile": percentile,
        "Threshold": threshold,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp,
        "Accuracy": accuracy_score(y_valid, predictions),
        "Precision": precision_score(
            y_valid, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_valid, predictions, zero_division=0
        ),
        "F1_Score": f1_score(
            y_valid, predictions, zero_division=0
        )
    })

autoencoder_results = pd.DataFrame(autoencoder_results)

display(autoencoder_results)

autoencoder_results.to_csv(
    r"D:\BATADAL\results\autoencoder_threshold_results.csv",
    index=False
)
```

```python
event_results = []

for attack_id, (start, end) in enumerate(
    test_attack_intervals,
    start=8
):
    start_time = pd.Timestamp(start)
    end_time = pd.Timestamp(end)

    event_mask = (
        (test_data["DATETIME"] >= start_time) &
        (test_data["DATETIME"] <= end_time)
    )

    event_predictions = test_predictions[event_mask]
    event_datetimes = test_data.loc[event_mask, "DATETIME"]

    detected_indices = np.where(event_predictions == 1)[0]
    detected = len(detected_indices) > 0

    if detected:
        first_detection = event_datetimes.iloc[detected_indices[0]]
        detection_delay = (
            first_detection - start_time
        ).total_seconds() / 3600
    else:
        first_detection = pd.NaT
        detection_delay = np.nan

    total_hours = int(event_mask.sum())
    detected_hours = int(event_predictions.sum())

    event_results.append({
        "Attack_ID": attack_id,
        "Start": start_time,
        "End": end_time,
        "Attack_Hours": total_hours,
        "Detected_Hours": detected_hours,
        "Coverage_Percent": (
            detected_hours / total_hours * 100
            if total_hours else 0
        ),
        "Event_Detected": detected,
        "First_Detection": first_detection,
        "Detection_Delay_Hours": detection_delay
    })

event_results = pd.DataFrame(event_results)

display(event_results.round(2))

event_results.to_csv(
    r"D:\BATADAL\results\autoencoder_event_detection_results.csv",
    index=False
)

print(
    "\nDetected attack events:",
    event_results["Event_Detected"].sum(),
    "out of",
    len(event_results)
)
```

```python
def create_sequences(data, window_size=24):
    sequences = []

    for i in range(len(data) - window_size + 1):
        sequences.append(data[i:i + window_size])

    return np.asarray(sequences, dtype=np.float32)


WINDOW_SIZE = 24  # previous 24 hours

# Normal Dataset 1 sequences
normal_sequences = create_sequences(
    X_train_scaled,
    WINDOW_SIZE
)

# Chronological 80/20 split
sequence_split = int(len(normal_sequences) * 0.80)

X_lstm_train = normal_sequences[:sequence_split]
X_lstm_validation = normal_sequences[sequence_split:]

# Dataset 2 and official test sequences
X_validation_sequences = create_sequences(
    X_valid_scaled,
    WINDOW_SIZE
)

X_test_sequences = create_sequences(
    X_test_scaled,
    WINDOW_SIZE
)

# A sequence label corresponds to its final hourly record
y_validation_sequences = y_valid.iloc[
    WINDOW_SIZE - 1:
].to_numpy()

y_test_sequences = y_test.iloc[
    WINDOW_SIZE - 1:
].to_numpy()

test_sequence_datetimes = test_data["DATETIME"].iloc[
    WINDOW_SIZE - 1:
].reset_index(drop=True)

print("LSTM training:", X_lstm_train.shape)
print("LSTM validation:", X_lstm_validation.shape)
print("Dataset 2 sequences:", X_validation_sequences.shape)
print("Test sequences:", X_test_sequences.shape)

print("\nValidation labels:", y_validation_sequences.shape)
print("Test labels:", y_test_sequences.shape)
```

```python
from tensorflow.keras.layers import (
    LSTM,
    RepeatVector,
    TimeDistributed,
    Dense
)
from tensorflow.keras.models import Sequential
from tensorflow.keras.callbacks import EarlyStopping

tf.keras.backend.clear_session()
tf.random.set_seed(42)
np.random.seed(42)

n_steps = X_lstm_train.shape[1]
n_features = X_lstm_train.shape[2]

lstm_autoencoder = Sequential([
    Input(shape=(n_steps, n_features)),

    LSTM(
        32,
        activation="tanh",
        return_sequences=False
    ),

    RepeatVector(n_steps),

    LSTM(
        32,
        activation="tanh",
        return_sequences=True
    ),

    TimeDistributed(
        Dense(n_features)
    )
])

lstm_autoencoder.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse"
)

lstm_early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    min_delta=0.0001
)

lstm_history = lstm_autoencoder.fit(
    X_lstm_train,
    X_lstm_train,
    validation_data=(
        X_lstm_validation,
        X_lstm_validation
    ),
    epochs=30,
    batch_size=64,
    shuffle=False,
    callbacks=[lstm_early_stopping],
    verbose=1
)

lstm_autoencoder.summary()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    lstm_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    lstm_history.history["val_loss"],
    label="Validation Loss"
)

plt.title("BATADAL LSTM Autoencoder Training")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    r"D:\BATADAL\results\lstm_autoencoder_training_loss.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Reconstruction error on normal validation sequences
lstm_normal_reconstruction = lstm_autoencoder.predict(
    X_lstm_validation,
    verbose=0
)

lstm_normal_errors = np.mean(
    np.square(
        X_lstm_validation -
        lstm_normal_reconstruction
    ),
    axis=(1, 2)
)

# Reconstruction error on official test sequences
lstm_test_reconstruction = lstm_autoencoder.predict(
    X_test_sequences,
    verbose=0
)

lstm_test_errors = np.mean(
    np.square(
        X_test_sequences -
        lstm_test_reconstruction
    ),
    axis=(1, 2)
)

lstm_results = []

for percentile in [95, 97.5, 99]:
    threshold = np.percentile(
        lstm_normal_errors,
        percentile
    )

    predictions = (
        lstm_test_errors > threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test_sequences,
        predictions,
        labels=[0, 1]
    ).ravel()

    lstm_results.append({
        "Threshold_Percentile": percentile,
        "Threshold": threshold,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp,
        "Accuracy": accuracy_score(
            y_test_sequences,
            predictions
        ),
        "Precision": precision_score(
            y_test_sequences,
            predictions,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test_sequences,
            predictions,
            zero_division=0
        ),
        "F1_Score": f1_score(
            y_test_sequences,
            predictions,
            zero_division=0
        )
    })

lstm_results = pd.DataFrame(lstm_results)

print(
    lstm_results.round(4).to_string(index=False)
)

lstm_results.to_csv(
    r"D:\BATADAL\results\lstm_autoencoder_threshold_results.csv",
    index=False
)
```

```python
# ---------- LSTM EVENT-LEVEL ANALYSIS ----------

lstm_event_results = []

for attack_id, (start, end) in enumerate(
    test_attack_intervals,
    start=8
):
    start_time = pd.Timestamp(start)
    end_time = pd.Timestamp(end)

    event_mask = (
        (test_sequence_datetimes >= start_time) &
        (test_sequence_datetimes <= end_time)
    ).to_numpy()

    event_predictions = final_lstm_predictions[event_mask]
    event_times = test_sequence_datetimes[event_mask]

    detected_positions = np.where(
        event_predictions == 1
    )[0]

    event_detected = len(detected_positions) > 0

    if event_detected:
        first_detection = event_times.iloc[
            detected_positions[0]
        ]

        delay_hours = (
            first_detection - start_time
        ).total_seconds() / 3600
    else:
        first_detection = pd.NaT
        delay_hours = np.nan

    total_hours = int(event_mask.sum())
    detected_hours = int(event_predictions.sum())

    lstm_event_results.append({
        "Attack_ID": attack_id,
        "Attack_Hours": total_hours,
        "Detected_Hours": detected_hours,
        "Coverage_Percent": (
            detected_hours / total_hours * 100
            if total_hours else 0
        ),
        "Event_Detected": event_detected,
        "First_Detection": first_detection,
        "Detection_Delay_Hours": delay_hours
    })

lstm_event_results = pd.DataFrame(
    lstm_event_results
)

display(lstm_event_results.round(2))

detected_events = int(
    lstm_event_results["Event_Detected"].sum()
)

print(
    f"LSTM detected {detected_events} out of "
    f"{len(lstm_event_results)} attack events."
)

lstm_event_results.to_csv(
    r"D:\BATADAL\results\lstm_event_results.csv",
    index=False
)


# ---------- FINAL THREE-MODEL COMPARISON ----------

final_comparison = pd.DataFrame([
    {
        "Model": "Isolation Forest",
        "Accuracy": accuracy_score(
            y_test, if_test_predictions
        ),
        "Precision": precision_score(
            y_test, if_test_predictions,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test, if_test_predictions,
            zero_division=0
        ),
        "F1_Score": f1_score(
            y_test, if_test_predictions,
            zero_division=0
        )
    },
    {
        "Model": "Dense Autoencoder",
        "Accuracy": accuracy_score(
            y_test, test_predictions
        ),
        "Precision": precision_score(
            y_test, test_predictions,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test, test_predictions,
            zero_division=0
        ),
        "F1_Score": f1_score(
            y_test, test_predictions,
            zero_division=0
        )
    },
    {
        "Model": "LSTM Autoencoder",
        "Accuracy": accuracy_score(
            y_test_sequences,
            final_lstm_predictions
        ),
        "Precision": precision_score(
            y_test_sequences,
            final_lstm_predictions,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test_sequences,
            final_lstm_predictions,
            zero_division=0
        ),
        "F1_Score": f1_score(
            y_test_sequences,
            final_lstm_predictions,
            zero_division=0
        )
    }
])

display(final_comparison.round(4))

final_comparison.to_csv(
    r"D:\BATADAL\results\final_three_model_comparison.csv",
    index=False
)

ax = final_comparison.set_index("Model").plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title("BATADAL Final Model Comparison")
plt.ylabel("Score")
plt.xlabel("Model")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.legend(loc="lower right")
plt.tight_layout()

plt.savefig(
    r"D:\BATADAL\results\final_three_model_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Final comparison saved successfully.")
```
